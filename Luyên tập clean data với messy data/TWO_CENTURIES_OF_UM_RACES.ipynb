```python
df2['month'] = df2['Event dates'].str.extract(r'\.(\d{2})\.\d{4}$')
```

Ý nghĩa từng phần:

r'' : viết raw string trong Python, để \d không bị hiểu nhầm thành ký tự đặc biệt.

\. : dấu chấm thật sự "." (phải thêm \ vì . trong regex nghĩa là “bất kỳ ký tự nào”).

(\d{2}) : nhóm 2 chữ số liên tiếp (chính là tháng cần lấy).

\. : lại là dấu ".".

\d{4} : 4 chữ số (năm, ví dụ 2018).

$ : kết thúc chuỗi (đảm bảo regex tìm ở cuối, ví dụ ...09.2020).
